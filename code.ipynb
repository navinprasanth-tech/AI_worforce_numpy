```python
import pandas as pd
```

```python
data=pd.read_csv("Iris (1).csv")
```

```python
# Task 1
import numpy as np

print("Shape of dataset:", data.shape)

print("Data types of each column:")
print(data.dtypes)

print("\nFirst 5 rows:")
print(data.head())
```

```python

# Task 2

numeric_data = data.iloc[:, :4].to_numpy()

mean_values = np.mean(numeric_data, axis=0)
median_values = np.median(numeric_data, axis=0)
std_values = np.std(numeric_data, axis=0)

print("Sepal Length: Mean =", mean_values[0],
      ", Median =", median_values[0],
      ", Std =", std_values[0])

print("Sepal Width: Mean =", mean_values[1],
      ", Median =", median_values[1],
      ", Std =", std_values[1])

print("Petal Length: Mean =", mean_values[2],
      ", Median =", median_values[2],
      ", Std =", std_values[2])

print("Petal Width: Mean =", mean_values[3],
      ", Median =", median_values[3],
      ", Std =", std_values[3])
```

```python

# Task 3

filtered_data = data[
    (data.iloc[:, 2] > 1.5) &
    (data.iloc[:, 1] < 3.0)
]

print("Filtered rows count:", len(filtered_data))

print("\nFirst 5 filtered rows:")
print(filtered_data.head())
```

```python

# Task 4: Min-Max Normalization

numeric_data = data.iloc[:, :4].to_numpy(dtype=float)

minimum = np.min(numeric_data, axis=0)
maximum = np.max(numeric_data, axis=0)

normalized_data = (numeric_data - minimum) / (maximum - minimum)

print("First 5 rows after normalization:")
print(np.round(normalized_data[:5], 3))
```

```python

# Task 5: Count Missing Values

numeric_data = data.iloc[:, :4].to_numpy(dtype=float)

missing_count = np.sum(np.isnan(numeric_data))

print("Total missing values:", missing_count)
```

```python

# Task 6: Replace Missing Values with Zero

numeric_data = data.iloc[:, :4].to_numpy(dtype=float)

data_with_zero = np.nan_to_num(numeric_data, nan=0.0)

print("Data after replacing missing values:")
print(data_with_zero[:5])
```
